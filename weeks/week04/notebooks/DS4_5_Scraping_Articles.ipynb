## Scraping the page

</br> With knowledge of the page structure, we can now proceed.

First, we'll download the HTML which defines the page, using the requests module and then create a BeautifulSoup representation of the page.

```python
# Scrape a specific article
url = 'https://www.economicsobservatory.com/from-crisis-to-stability-what-next-for-bolivias-economy'
headers = {'User-Agent': 'PP434 course notebook (LSE)'}   # say who we are (some sites block anonymous requests)
response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, 'html.parser')
```

</br></br>

Next we can get the article title, which we identified as having a class `article_title`:

```python

# Get the article title
article_title = soup.find('h1', class_='article__title').get_text()
print(f"Article Title: {article_title}")
```

</br></br>

We can do the same for the category:

```python

# Get the article category
category = soup.find('a', class_='primary-category').get_text()
print(f"Category: {category}")
```

</br></br>

... and then extract the article's full text:

```python


# Get the entire article text
article_text = soup.find('article', class_='article__content').get_text(separator=' ')
print(f"\nArticle length: {len(article_text)} characters")
```

## Word Cloud Visualization

Let's visualise the text in this article using a word cloud, using `WordCloud` and `Matplotlib`

```python
# Install packages: wordcloud and matplotlib
%pip install wordcloud matplotlib --quiet
```

```python
# Import the packages we just installed
from wordcloud import WordCloud
import matplotlib.pyplot as plt
```

```python
# Create and display the word cloud
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(article_text)

plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title(f'Word Cloud for: {article_title}', fontsize=14, pad=20)
plt.show()
```

**Note**: The WordCloud module automatically removes common stopwords like "and," "the," and "of." It uses a predefined list of stopwords to filter out these high-frequency, low-value words by default. We could also add or customise stopwords if we want to remove additional terms specific to our content.
