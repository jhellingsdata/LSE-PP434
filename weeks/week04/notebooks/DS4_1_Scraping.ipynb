## Searching for elements
`BeautifulSoup` lets us search for elements on the page.  We can do this in at least three ways: </br></br>

- **`id`**. The unique identifier given to elements.
    
    For example we can access a figure: `<figure id="chart1">` with `soup.find(id="chart1")` </br></br></br>



- **`class`**. The class attribute is used for styling multiple elements.

    For example we can find elements with class "big": `<h1 class="big">` with `soup.find(class_="big")`</br></br></br>

- **Element type**. The HTML tag name itself.

    For example we can find all heading elements: `<h2>` with `soup.find_all("h2")`

<br>
<br>

Let's try each. We can use these methods with `BeautifulSoup`'s `find`, which returns a single element, or `find_all` which returns a list:

</br></br></br>

First, let's try finding by `id`. `id`s should be unique, so we'll use `find`:

```python
chart1 = soup.find(id="Location1")
chart1
```

</br></br>

Next, let's find all the chart descriptions. These each have a `class` of `chart-description`.

```python
chart_descriptions = soup.find_all(class_="chart-description")
chart_descriptions
```

<br>
<br>

Finally, we can search just by element type. For example, we could find the page title: contained in an `<h1>`:
<br>
<br>

```python
soup.find("h1")
```

</br></br>

Or find every `<figure>`:

```python
soup.find_all("figure")
```

<br>
<br>

## Accessing properties

We've located the elements, but we're getting the full html representation back. We want to be able to extract properties of them such as their text content or attributes like `id`, `class`, or `href`.

<br>
<br>

First, let's get just the text content of the page's main heading:

```python
page_title = soup.find("h1")
print(page_title.text)
```

<br>
<br>

Next, let's loop through all the figures and extract their `id` attributes:

```python
figures = soup.find_all("figure")
for figure in figures:
    print(figure.get("id"))
```

# Putting this all together

Let's put together every function we've seen, so that we can generate a report on a portfolio page.

We'll:

1. Download the page's HTML.
2. Parse the page with `BeautifulSoup`.
3. Find the main heading, and extract its text.
4. Find every chart, and every chart description.

```python
# 1. Downloading the page's HTML.
portfolio_url = "http://mclass-user.github.io"
response = requests.get(portfolio_url)
html_content = response.content

# 2. Parsing the HTML with BeautifulSoup.
soup = BeautifulSoup(html_content, "html.parser")

#3. Find the main heading and get its text content.
main_heading_element = soup.find("h1")
main_heading_text = main_heading_element.get_text(strip=True)
print("Page heading:", main_heading_text)

print("")

#4. Find all charts and chart titles
chart_elements = soup.find_all("figure")
print("Found", len(chart_elements), "charts. With ids:")
for chart in chart_elements:
    print("-", chart.get("id"))

print("")

# ...and the titles:
chart_description_elements = soup.find_all(class_="chart-description")
print("Found", len(chart_description_elements), "chart descriptions:")
for chart_description_element in chart_description_elements:
    print("")
    print("\t"+chart_description_element.get_text(strip=True)) # The \t gives us a tab character to indent the text
    print("")

```
