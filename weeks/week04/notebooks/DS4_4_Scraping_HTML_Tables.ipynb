Before starting, we have to install and import the tool Pandas, which we'll use for scraping and data manipulation

```python
# pandas is already installed in Colab, so there is nothing to install here
```

```python
import pandas as pd # used for data scraping, and manipulation
```

First we define the URL of web-page we want to scrape from.


> **2026 update:** Wikipedia now blocks requests from code that doesn't identify itself (you'd see `HTTPError: HTTP Error 403: Forbidden`). We fix this by sending a `User-Agent`: a short label saying who is asking.

```python
url = "https://en.wikipedia.org/wiki/G7"   # The URL of the webpage

# Wikipedia blocks requests that don't say who they are, so we send a "User-Agent" header (a short description of us).
# match= keeps only the tables that contain this text, so we don't have to search through every table on the page.
tables_from_webpage = pd.read_html(url, match='Location held', storage_options={'User-Agent': 'PP434 course notebook (LSE)'})

len(tables_from_webpage)   # how many tables matched
```

`pd.read_html` returns a **list** of tables (DataFrames), one for each table it finds.

Because we used `match='Location held'`, the list only contains the summits table, so it's the first item: `[0]`.

```python
tables_from_webpage[0].head(10)
```

Let's call this table `df`, for ease of use. Websites change, so check the column names match what we use below:

```python
df = tables_from_webpage[0]
df.columns
```

## Manipulating the Data

Let's count how many summits each country has hosted.

The table has a few **label rows** (e.g. "As the G7") that stretch across the whole row, so the same text appears in every column. We remove those first.

```python
df = df[df['Host'] != df['#']]          # remove label rows (where every column has the same text)

counts = df['Host'].value_counts()       # count how many times each country appears
counts = counts.reset_index()            # turn it back into a table, with columns 'Host' and 'count'

counts
```

## Uploading the Data

Now let's save our table to upload to GitHub and use in Vega-lite

```python
counts.to_csv('g7_summits.csv', index=False) # Save the data to a CSV file
```
