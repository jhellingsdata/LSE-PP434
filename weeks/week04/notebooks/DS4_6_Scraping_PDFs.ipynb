# PDF Data Extraction: Two Approaches

PDFs come in two main types when it comes to data extraction:

1. **Text-based PDFs**: Created digitally (from Word, Excel, etc.) - text is selectable and extractable
2. **Image-based PDFs**: Scanned documents where text is "burned in" as images

We'll demonstrate the first case, with a recent UK government evaluation as an example, and provide resources for the second.


## Part 1: Text-Based PDFs with Camelot

When PDFs contain actual text (not images of text), we can extract tables directly using `camelot-py`. This is faster and more accurate than OCR.

First, let's install and import what we need:

```python
import camelot
import pandas as pd
import matplotlib.pyplot as plt
```

### Extracting Tables from Elizabeth Line Report

The Elizabeth Line evaluation report contains tables with transport data. Let's extract them:

```python
# Load the Elizabeth Line PDF
# The PDF is stored in the course repository. camelot can read it straight from the link.
pdf_path = "https://raw.githubusercontent.com/jhellingsdata/LSE-PP434/main/weeks/week04/data/elizabeth-line-evaluation_excerpt.pdf"

# Extract all tables using the 'lattice' method (good for tables with clear borders)
tables = camelot.read_pdf(pdf_path, flavor='lattice')

print(f"Found {len(tables)} tables in the PDF")

# Show basic info about each table
for i, table in enumerate(tables):
    print(f"Table {i+1}: {table.shape[0]} rows × {table.shape[1]} columns")
    print(f"  Parsing confidence: {table.accuracy:.1f}%")
```

```python
df = tables[0].df
print("Raw extracted table:")
print(df.head())

# Basic cleaning - remove empty rows/columns
df_clean = df.dropna(how='all').dropna(axis=1, how='all')

print(df_clean.head())
```

### Alternative: Stream Method

If the 'lattice' method doesn't work well (e.g., tables without clear borders), try the 'stream' method:

```python
# Try the stream method as an alternative
tables_stream = camelot.read_pdf(pdf_path, flavor='stream')
print(f"Stream method found {len(tables_stream)} tables")

if len(tables_stream) > 0:
    df_stream = tables_stream[0].df
    print("First table using stream method:")
    print(df_stream.head())
```

## Part 2: Image-Based PDFs with OCR

When PDFs are scanned documents or contain images of text, we need **Optical Character Recognition (OCR)** to extract the text. This is significantly more complex than text-based extraction and often requires manual cleanup of results, and beyond the scope of this course.


**Image-based PDFs** occur when:
- Documents are scanned from physical papers
- PDFs are created from photographs  
- Text is embedded as images rather than selectable text
- You can't highlight/copy text from the PDF

**The OCR Process:**
1. **PDF → Images**: Convert each PDF page to an image file
2. **Images → Text**: Use OCR software to "read" text from images  
3. **Text → Data**: Clean and structure the raw OCR output

### Tools and Resources for OCR

If you need to work with scanned PDFs, here are the main approaches:

**Python Libraries:**
- `pytesseract` (Tesseract OCR wrapper) - free, open-source
- `pdf2image` (converts PDF pages to images)
- `camelot` with `cv2` backend (can handle some image-based tables)

**Commercial Solutions:**
- Adobe Acrobat Pro (built-in OCR)
- ABBYY FineReader (specialized OCR software)
- Google Cloud Vision API / AWS Textract (cloud-based)
