# Loops and conditionals: quick start

Two ideas that let code do repetitive work for us:

- **Loops** repeat the same steps for every item in a list (every country, every data series, every chart)
- **Conditionals** (`if` / `else`) let code make a decision

## 1. Why loops?

Say we want to print a message for three cities. Without a loop, we copy the same line again and again:

```python
print("Data for London")
print("Data for Cardiff")
print("Data for Belfast")
```

This works for three cities. For 50 countries it's slow to write and easy to get wrong. A **loop** does it for us.

## 2. Your first loop

First, a **list** (from Week 3):

```python
cities = ["London", "Cardiff", "Belfast"]
```

A `for` loop runs the indented lines **once for each item** in the list. Each time round, `city` holds the next item:

```python
for city in cities:
    print(city)
```

<br>

Three things to notice:

| Part | Meaning |
|---|---|
| `for city in cities:` | Go through `cities` one item at a time, calling the current item `city` |
| `:` | Ends the `for` line (easy to forget!) |
| Indent (4 spaces) | The lines that repeat. Anything not indented runs once, after the loop |

The name `city` is our choice: `for x in cities:` would work the same way. A descriptive name makes the code easier to read.

```python
for city in cities:
    print("Starting:", city)
    print("Finished:", city)

print("All done")   # not indented, so runs once at the end
```

```python
for x in cities:    # The variable name can be anything
    print("Starting:", x)
    print("Finished:", x)

print("All done")   # not indented, so runs once at the end
```

## 3. Putting values into text: f-strings

We often want to put a value **inside** some text. Put an `f` before the quotes, and the variable name in `{ }`:

```python
city = "London"

f"Data for {city}"
```

We can put in **as many values as we like**, each in its own `{ }`:

```python
team = "Wales"
player = "Gareth Bale"

f"The best football team in the world is {team}, and the best player is {player}"
```

We can even do a small calculation inside the `{ }`:

```python
f"{city} has {len(city)} letters"
```

Combined with a loop, we can make a different message for each item:

```python
for city in cities:
    print(f"Data for {city}")
```

<br>

### Building web addresses in a loop

This is where loops become useful for data. FRED (the US Federal Reserve's database) gives each data series a code, and each code has its own web page. Let's make a link for each one:

```python
fred_series = ["UNRATE", "CPIAUCSL", "GDPC1"]   # unemployment, prices, real GDP

for series_id in fred_series:
    url = f"https://fred.stlouisfed.org/series/{series_id}"
    print(url)
```

Click the links to check they work. In **DS5_2**, we use exactly this pattern to **download** each series.

## 4. Conditionals: `if` and `else`

A conditional runs code **only if** something is true:

```python
inflation = 3.8

if inflation > 2:
    print("Above the 2% target")
else:
    print("At or below the 2% target")
```

<br>

Like `for`, the `if` and `else` lines end with `:`, and the lines underneath are indented.

Comparisons we can use:

| Code | Meaning |
|---|---|
| `==` | equal to (two `=`: one `=` *stores* a value) |
| `!=` | not equal to |
| `>`, `<` | greater than, less than |
| `>=`, `<=` | greater than or equal to, less than or equal to |

Try changing `inflation` above to `1.5` and running it again.

## 5. A conditional inside a loop

Now combine the two: check **every** value in a list.

```python
inflation_rates = [3.8, 1.7, 2.0, 5.2]

for rate in inflation_rates:
    if rate > 2:
        print(f"{rate}% is above target")
    else:
        print(f"{rate}% is at or below target")
```

Notice the two levels of indentation: the `if`/`else` is inside the loop, and each `print` is inside the `if` or `else`.

## Annex: the older `.format()` method

You'll often see another way of putting values into text in older code (including some of our examples): **`.format()`**.

The text has empty `{ }` placeholders, and `.format()` fills them **in order**:

```python
sentence = "The best football team in the world is {}, and the best player is {}"

sentence.format("Wales", "Gareth Bale")
```

<br>

**Why does it still exist?** With `.format()`, the text is written **first** and filled in **later**. This makes it handy for a *template* we define once at the top of a notebook and reuse, for example a web address:

```python
url_template = "https://fred.stlouisfed.org/series/{}"     # defined once, e.g. at the top of a notebook

for series_id in fred_series:
    print(url_template.format(series_id))
```

<br>

An f-string can't do this: it fills in the values **immediately**, so the variables must already exist on that line.

> **Careful with numbered placeholders.** `{0}`, `{1}`, ... pick values by position, and Python counts from 0. What does this print, and why?

```python
"The best team is {1} and the best player is {2}".format("Wales", "Gareth Bale", "London")
```
