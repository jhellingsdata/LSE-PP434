# Downloading many data series with a loop (clean version)

The full process from **DS5_2_Loop_API_Download**, without the explanations. Add your series codes to the list `fred_series` and run all the cells.

> 🔑 This uses the shared course API key. Never put your own key in a public repository or website (see DS5_2 for how to keep it secret).

```python
import requests

api_key = "22ee7a76e736e32f54f5df0a7171538d"
fred_series = []      # your series codes, e.g. ["UNRATE", "CPIAUCSL", ...]
```

```python
for series_id in fred_series:
    print(f"--- Series: {series_id} ---")

    url = f"https://api.stlouisfed.org/fred/series/observations?series_id={series_id}&api_key={api_key}&file_type=json"
    response = requests.get(url)
    print(f"Response: {response.status_code}")

    if response.status_code == 200:
        filename = f"data_US_FRED-{series_id}.json"
        with open(filename, "w") as file:
            file.write(response.text)
        print(f"OK: saved as {filename}")
    else:
        print(f"Problem with {series_id}: nothing saved")
    print()

print("Finished")
```
