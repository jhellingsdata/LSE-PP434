# Making many charts with a loop

A Vega-Lite chart is just a **dictionary** of settings. To make the same chart for another country, we only need to change the **data link** and the **title**. A loop can do that for every country in a list.

```python
import requests        # fetch things from the internet
import altair as alt    # show and save Vega-Lite charts
```

<br><br>

---

<br>

## 1. A chart is a dictionary

Let's fetch a simple line chart from Richard's chart library. Because it's JSON, `.json()` turns it into a Python dictionary:

```python
url = "https://raw.githubusercontent.com/RDeconomist/RDeconomist.github.io/main/charts/library/chartLine0.json"

spec = requests.get(url).json()
spec
```

We can read and change any part using keys, just like in Week 3:

```python
spec["title"]["text"]
```

```python
spec["title"]["text"] = "My new title"
spec["width"] = 400
spec["$schema"] = "https://vega.github.io/schema/vega-lite/v6.json"   # the latest Vega-Lite version, to match our websites

alt.Chart.from_dict(spec)    # show the dictionary as a chart
```

<br><br>

---

<br>

## 2. Point it at new data: one country

The Economics Observatory API (Week 4) has the same series for many countries. Only the **country code** in the link changes:

`https://api.economicsobservatory.com/gbr/unem?vega`

This data has columns called `date` and `value`, so we update the chart to use them. As always, we get it working for **one** country first:

```python
spec["encoding"]["x"]["field"] = "date"
spec["encoding"]["y"]["field"] = "value"

country = "gbr"
spec["data"]["url"] = f"https://api.economicsobservatory.com/{country}/unem?vega"
spec["title"]["text"] = f"Unemployment: {country}"

alt.Chart.from_dict(spec)
```

<br><br>

---

<br>

## 3. The loop: one chart per country

The same lines, inside a loop. Each time round we change the link and the title, show the chart, and **save** it as a JSON file for our website:

```python
countries = ["gbr", "usa", "can", "egy"]

for country in countries:
    spec["data"]["url"] = f"https://api.economicsobservatory.com/{country}/unem?vega"
    spec["title"]["text"] = f"Unemployment: {country}"

    chart = alt.Chart.from_dict(spec)
    chart.display()                                # show it in the notebook
    chart.save(f"unemployment_{country}.json")     # save it for our website
```

<br>

> `chart.display()` is needed inside a loop: a notebook only shows the **last** result of a cell automatically.

<br><br>

---

<br>

## 4. Add a conditional: highlight one country

Say we want the UK in red and every other country in grey. That's a decision for each chart, so we use `if`/`else` inside the loop:

```python
for country in countries:
    spec["data"]["url"] = f"https://api.economicsobservatory.com/{country}/unem?vega"
    spec["title"]["text"] = f"Unemployment: {country}"

    if country == "gbr":
        spec["mark"]["color"] = "red"
    else:
        spec["mark"]["color"] = "grey"

    chart = alt.Chart.from_dict(spec)
    chart.display()
    chart.save(f"unemployment_{country}.json")
```
