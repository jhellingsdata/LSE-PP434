## 2. A first chart

An Altair chart has three parts, the same as Vega-Lite:

| Altair | Vega-Lite | Meaning |
|---|---|---|
| `alt.Chart(df)` | `"data"` | Which data |
| `.mark_line()` | `"mark": "line"` | What shape to draw |
| `.encode(x=..., y=...)` | `"encoding"` | Which columns go where |

The letter after the colon is the **type**: `:T` for time (temporal), `:Q` for numbers (quantitative), `:N` for text (nominal), `:O` for ordered text (ordinal).

```python
chart = alt.Chart(df).mark_line().encode(
    x=alt.X('Date_iso:T'),
    y=alt.Y('PE_ratio:Q')
)

chart
```

<br>
<br>

### Under the hood: it's Vega-Lite

Altair has written a Vega-Lite spec for us. `.to_dict()` gives it to us as a Python dictionary, so we can look inside with keys:

```python
spec = chart.to_dict()

spec['mark']
```

```python
spec['encoding']
```

<br>

Recognise it? `mark`, `encoding`, `field`, `type`: exactly what we edited by hand in Week 2.

<br>
<br>

### What's that spike?

In 2009, during the financial crisis, company **earnings collapsed** almost to zero. Dividing by a tiny number gives a huge P/E, even though share prices *fell*.

This is one reason economists prefer a smoothed measure (see the CAPE stretch exercise at the end). It's also a good reminder to ask what's driving a pattern before you write a headline about it.

## 3. Style it

The same styling ideas as Week 2, written in Python. Change one thing at a time and re-run the cell.

| Week 2 (JSON) | Altair | Documentation |
|---|---|---|
| `"title": {...}` | `.properties(title=alt.Title(...))` | [Adjusting the title](https://altair-viz.github.io/user_guide/customization.html#adjusting-the-title) |
| `"width": 400`, `"height": 250` | `.properties(width=400, height=250)` | [Adjusting chart size](https://altair-viz.github.io/user_guide/customization.html#adjusting-chart-size) |
| `"mark": {"color": ...}` | `.mark_line(color=...)` | [Line mark](https://altair-viz.github.io/user_guide/marks/line.html) |
| `"mark": {"strokeWidth": 2}` | `.mark_line(strokeWidth=2)` | [Line mark](https://altair-viz.github.io/user_guide/marks/line.html) |
| `"axis": {"title": null}` | `alt.X(...).title(None)` | [Adjusting axis labels](https://altair-viz.github.io/user_guide/customization.html#adjusting-axis-labels) |
| `"axis": {"grid": false}` | `alt.X(...).axis(grid=False)` | [All axis options](https://altair-viz.github.io/user_guide/generated/core/altair.Axis.html) |
| `"axis": {"labelFontSize": 12}` | `alt.X(...).axis(labelFontSize=12)` | [All axis options](https://altair-viz.github.io/user_guide/generated/core/altair.Axis.html) |

> Notice the pattern: things that apply to the **whole chart** (title, size) go in `.properties()`, like the top level of a Vega-Lite spec.

```python
chart_styled = alt.Chart(df).mark_line(
    color='#c8102e',
    strokeWidth=1.5
).encode(
    x=alt.X('Date_iso:T').title(None).axis(grid=False),
    y=alt.Y('PE_ratio:Q').title(None)
).properties(
    width=500,
    height=280,
    title=alt.Title(
        text='US stock prices are high relative to earnings',
        subtitle='Price-to-earnings ratio, S&P composite. Source: Robert Shiller',
        anchor='start'
    )
)

chart_styled
```

## 4. Save the chart for your website

So far, Altair has copied all the data **into** the chart. For a website, it's better to **link** to the data instead, just like in Week 2 (`"data": {"url": ...}`). The chart file stays small, and it updates when the data does.

We do this by giving `alt.Chart()` the **URL** instead of the table. With a URL, Altair can't see the columns, so the types (`:T`, `:Q`) are required.

```python
web_chart = alt.Chart(data_url).mark_line(
    color='#c8102e',
    strokeWidth=1.5
).encode(
    x=alt.X('Date_iso:T').title(None).axis(grid=False),
    y=alt.Y('PE_ratio:Q').title(None)
).properties(
    width=500,
    height=280,
    title=alt.Title(
        text='US stock prices are high relative to earnings',
        subtitle='Price-to-earnings ratio, S&P composite. Source: Robert Shiller',
        anchor='start'
    )
)

web_chart
```

Save it as a JSON file:

```python
web_chart.save('pe_ratio_chart.json')
```

```python
# Check: the data is a link, not thousands of rows
print(web_chart.to_json()[:400])
```

## 6. Optional: add the long-run average (layering charts)

Is today's P/E *high*? Only compared with something. We can add a dashed line at the long-run average.

This uses two ideas we'll cover properly in a later week, but they show how easily charts combine in Altair:
- `mean(PE_ratio)` asks Altair to **calculate** the average for us
- `+` **layers** one chart on top of another

> Read more: [Layered charts](https://altair-viz.github.io/user_guide/compound_charts.html#layered-charts)

```python
average_line = alt.Chart(df).mark_rule(
    color='grey',
    strokeDash=[4, 4]
).encode(
    y=alt.Y('mean(PE_ratio):Q')
)

chart_styled + average_line
```

<br>

Now the chart shows the comparison the title talks about: the P/E today is well above its long-run average.

To save the layered version for your website, make `average_line` with `data_url` instead of `df`, then save `web_chart + average_line`.

> **Note:** this is a simple P/E ratio. The famous "Shiller P/E" (or CAPE) divides by the average of the last 10 years of earnings, to smooth out booms and recessions. See the stretch exercise below.
