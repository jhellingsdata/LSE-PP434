## 2. Variables

A variable is a **name** that stores a value, so we can use it later. We create one with `=`.

```python
price = 120
earnings = 8

ratio = price / earnings
ratio
```

The last line of a cell is shown automatically. Use `print()` to show more than one thing.

```python
print(price)
print(ratio)
```

## 3. Types of data

Every value has a **type**. The three you will see most:

| Type | Example | What it is |
|---|---|---|
| `int` | `120` | Whole number |
| `float` | `15.0` | Number with a decimal point |
| `str` | `"UK"` | Text (a *string*), always in quotes |

Check a type with `type()`:

```python
type(ratio)
```

```python
country = "UK"
type(country)
```

> **Why it matters:** `"120"` (text) and `120` (a number) look the same, but you can only do maths with one of them. We'll meet this problem with dates in the practical.

## 4. Lists

A **list** stores several values, in order, inside square brackets `[ ]`.

```python
cities = ["London", "Cardiff", "Edinburgh", "Belfast"]
cities
```

Get an item by its position. **Python counts from 0**, so the first item is `[0]`.

```python
cities[0]
```

```python
cities[-1]   # negative numbers count from the end
```

## 5. Dictionaries (this is JSON!)

A **dictionary** stores **key: value** pairs inside curly brackets `{ }`.

This should look familiar: it's the same structure as the Vega-Lite JSON you edited in Week 2.

```python
chart = {
    "mark": "line",
    "width": 400,
    "encoding": {
        "x": {"field": "date", "type": "temporal"},
        "y": {"field": "cpih", "type": "quantitative"}
    }
}
chart
```

Get a value by its **key**:

```python
chart["mark"]
```

Keys can be chained to reach inside nested dictionaries:

```python
chart["encoding"]["y"]["field"]
```

And we can change a value, just like editing the JSON by hand:

```python
chart["mark"] = "bar"
chart["mark"]
```
